```python
import uiautomator2 as u2
import subprocess
import time
import os

d = u2.connect()

def adb(cmd):
    subprocess.run(['adb','shell'] + cmd.split())

def wait(sec):
    time.sleep(sec)

def typ(text):
    subprocess.run(['adb','shell','am','broadcast','-a','ADB_INPUT_TEXT','--es','msg',text])
    wait(0.5)

print("="*50)
print("NEURODROID BCI SYSTEM")
print("DigitalOcean Gradient Hackathon")
print("Helping Disabled People!")
print("Built by Sachin Sharma")
print("="*50)

rec = subprocess.Popen([
    'adb','shell','screenrecord',
    '--size','720x1600',
    '--bit-rate','8000000',
    '--time-limit','120',
    '/sdcard/final_demo.mp4'
])
wait(2)
print("Recording Started!")

print("\nUnlocking Phone...")
adb('input keyevent KEYCODE_WAKEUP')
wait(1)
d.swipe(0.5,0.85,0.5,0.4,duration=0.5)
wait(3)
print("  Face Unlock Attempted...")
adb('input text 1111')
wait(0.5)
adb('input keyevent KEYCODE_ENTER')
wait(2)
print("Phone Unlocked!")

print("\nCalling +918906233009...")
adb('am start -a android.intent.action.CALL -d tel:+918906233009')
wait(2)
print("  Calling now...")
wait(10)
adb('input keyevent KEYCODE_ENDCALL')
wait(2)
d.press('home')
wait(2)
print("Call Done!")

print("\nOpening SMS...")
adb('am start -a android.intent.action.SENDTO -d sms:+918906233009')
wait(3)
if d(className='android.widget.EditText').exists(timeout=3):
    d(className='android.widget.EditText').click()
    wait(1)
    typ('Hello')
    wait(1)
if d(resourceId='com.google.android.apps.messaging:id/send_message_button').exists(timeout=2):
    d(resourceId='com.google.android.apps.messaging:id/send_message_button').click()
elif d(description='Send').exists(timeout=2):
    d(description='Send').click()
else:
    adb('input keyevent KEYCODE_ENTER')
wait(1)
d.press('home')
wait(1)
print("SMS Done!")

print("\nOpening WhatsApp...")
d.app_start('com.whatsapp')
wait(3)
if d(description='Search').exists(timeout=2):
    d(description='Search').click()
wait(1)
typ('8906233009')
wait(1)
try:
    d(textContains='8906').click()
except:
    d(resourceId='com.whatsapp:id/contact_row_container').click()
wait(1)
if d(resourceId='com.whatsapp:id/entry').exists(timeout=2):
    d(resourceId='com.whatsapp:id/entry').click()
    wait(0.5)
    typ('Hello')
    wait(0.5)
if d(resourceId='com.whatsapp:id/send').exists(timeout=1):
    d(resourceId='com.whatsapp:id/send').click()
elif d(description='Send').exists(timeout=1):
    d(description='Send').click()
wait(1)
d.press('home')
wait(1)
print("WhatsApp Done!")

print("\nOpening Instagram Reels...")
d.app_start('com.instagram.android')
wait(5)
if d(description='Reels').exists(timeout=3):
    d(description='Reels').click()
    wait(3)
elif d(description='Reels tab').exists(timeout=2):
    d(description='Reels tab').click()
    wait(3)
for i in range(3):
    wait(3)
    d.swipe(0.5,0.75,0.5,0.25,duration=0.5)
    wait(2)
    print(f"  Reel {i+1}/3!")
d.press('home')
wait(2)
print("Instagram Done!")

print("\nOpening YouTube Shorts...")
d.app_start('com.google.android.youtube')
wait(5)
if d(description='Shorts').exists(timeout=3):
    d(description='Shorts').click()
    wait(3)
    d.click(0.5,0.5)
    wait(10)
else:
    if d(description='Search YouTube').exists(timeout=2):
        d(description='Search YouTube').click()
    elif d(description='Search').exists(timeout=2):
        d(description='Search').click()
    wait(1)
    typ('YouTube Shorts')
    wait(1)
    adb('input keyevent KEYCODE_ENTER')
    wait(3)
    d.click(0.5,0.25)
    wait(10)
d.press('home')
wait(2)
print("YouTube Shorts Done!")

subprocess.run(['adb','shell','pkill','-2','screenrecord'])
wait(5)
subprocess.run(['adb','pull','/sdcard/final_demo.mp4'])
wait(2)
size = os.path.getsize('final_demo.mp4')
print(f"Video Size: {size/1024/1024:.1f} MB")
if size > 1:
    print("VIDEO SAVED!")
    print("File: final_demo.mp4")
else:
    print("File too small!")

print("="*50)
print("DEMO COMPLETE!")
print("Unlock       - Done!")
print("Call         - Done!")
print("SMS          - Done!")
print("WhatsApp     - Done!")
print("Instagram    - Done!")
print("YouTube      - Done!")
print("="*50)
print("NeuroDroid BCI System")
print("DigitalOcean Hackathon")
print("Built by Sachin Sharma!")
```
